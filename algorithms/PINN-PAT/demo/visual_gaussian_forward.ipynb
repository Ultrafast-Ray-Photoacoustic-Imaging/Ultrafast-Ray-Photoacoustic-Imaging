#### Visualization demo of `pinn_pat_gs_init.ckpt`

- Checkpoint is generated by `../run/run_gaussian_forward.py`

- The PINN with gaussian-shaped initial pressure is trained for only 50000 iteration. It can be used as a pretrained model for PAT reconstruction optimization to accelerate covergence and improve  accuracy.

------

```python
import torch
torch.set_grad_enabled(False)
import sys
from matplotlib import pyplot as plt
import json
from pprint import pprint
sys.path.append("..")


from models.pinn_for_pat import PINNForPAT
from models.arg import Arg
device = torch.device("cpu")
```

```python
ckpt_filepath = "../ckpt/gaussian/pinn_pat_gs_init.ckpt"
state_dict = torch.load(ckpt_filepath, map_location=torch.device('cpu'))

arg_dict = json.loads(bytes(state_dict['arg'].numpy()).decode('utf-8'))
pprint(arg_dict)

arg = Arg.from_dict(arg_dict)
arg.load_filepath = ckpt_filepath
pinn_pat = PINNForPAT(arg, device=device)
```

### Pressure field evolution (0~1)

```python
H = 256
X_axis = torch.linspace(0, 1, 2 * H + 1)[1::2]
x, y = torch.meshgrid([X_axis] * 2, indexing="ij")

fig, axes = plt.subplots(dpi=100, figsize=(20, 2), ncols=11)
list(map(lambda ax:ax.axis(False), axes.flatten()))
for it, t in enumerate(torch.linspace(0, 1, 11)):
    p0 = pinn_pat.p(torch.stack([x, y, torch.zeros_like(x) + t], dim=-1))
    if it == 0:
        vmax = max(p0.max(), 1e-5)
    axes[it].imshow(p0, cmap="bwr", vmax=vmax, vmin=-vmax)
    axes[it].set_title(f"t={t:.1f}")
    
```
